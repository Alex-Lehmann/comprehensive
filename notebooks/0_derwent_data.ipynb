Bringing in the fields from the website.

```python
path = '../.data/derwent/ai' 

ai = pd.read_pickle(f'{path}/ai_api.pkl') # These might not run sequentially
ai.rename(columns={'cdpt_dwpi': 'n_backward_citations',
                   'cgpt_dwpi': 'n_forward_citations',
                   'co_dwpi': 'assignee',
                   'pd_orig': 'publication_date',
                   'pn_orig': 'publication_number',
                   'ref_derwent': 'family_root'},
          inplace=True)

dl = []
for f in os.listdir(f'{path}/dl'):
    df = pd.read_excel(f'{path}/dl/{f}', skiprows=1)
    df.rename(columns={'Publication Number': 'publication_number',
                       'IPC Section - DWPI': 'ipc_section',
                       'IPC Class - DWPI': 'ipc_class',
                       'IPC Subclass - DWPI': 'ipc_subclass',
                       'IPC Class Group - DWPI': 'ipc_class_group',
                       'IPC Subgroup - DWPI': 'ipc_subgroup',
                       'Abstract (English)': 'abstract',
                       'Claims Count': 'n_claims',
                       'DWPI Count of Family Members': 'family_size',
                       'Cited Refs - Patent': 'backward_citations',
                       'Strategic Importance': 'strategic_importance',
                       'Probability of Grant': 'p_grant',
                       'Domain Influence': 'domain_influence',
                       'Combined Patent Impact': 'combined_patent_impact'},
              inplace=True)
    dl.append(df)
dl = pd.concat(dl)
dl.dropna(subset=['ipc_class', 'ipc_class_group', 'ipc_section', 'ipc_subclass',
                  'ipc_subgroup', 'abstract', 'n_claims', 'family_size',
                  'strategic_importance', 'p_grant', 'domain_influence',
                  'combined_patent_impact'],
          inplace=True)

dl.abstract = dl.abstract.map(lambda x: regex.sub('\n',
                                                  ' ',
                                                  x,
                                                  flags=regex.IGNORECASE))
dl.ipc_subgroup = dl.ipc_subgroup.map(lambda x: regex.sub(r'\s+\(.*?\)', '', x))

dl.ipc_section = dl.ipc_section.map(lambda x: x.split(', '))
dl.ipc_class = dl.ipc_class.map(lambda x: x.split(', '))
dl.ipc_subclass = dl.ipc_subclass.map(lambda x: x.split(', '))
dl.ipc_class_group = dl.ipc_class_group.map(lambda x: x.split(', '))
dl.ipc_subgroup = dl.ipc_subgroup.map(lambda x: x.split(', '))
dl.backward_citations = dl.backward_citations.map(lambda x: x.split(' | '),
                                                  na_action='ignore')

dl.p_grant = dl.p_grant / 100

ai = pd.merge(ai, dl, on='publication_number')
ai.replace({np.nan: None}, inplace=True)

ai.to_pickle(f'{path}/ai_patents.pkl')

ai
```

# CCMT

```python
ccmt_search = SearchByQuery(
    'cpc=(Y02B OR Y02C OR Y02D OR Y02E OR Y02T OR Y04S) AND pd>=(20000101) AND pd<(20200101) NOT ki=(U* OR Y* OR Z*)',
    ['pn', 'pd', 'cdpt', 'cgpt', 'co'],
    size=1000000,
    search_limit=1000000,
    display_limit=1000000,
    list_filter='collapse:derwent-family'
)
ccmt_patents = ccmt_search.post()
ccmt_families = ccmt_patents[ccmt_patents.ref_derwent.isnull()]

print(f'# of patents: {len(pd.unique(ccmt_patents.id))}')
print(f'# of families: {len(pd.unique(ccmt_families.id))}')

for i, batch in enumerate(batched(ccmt_patents.pn_orig, 30000)):
    with open(f'../.data/derwent/ccmt_pns_{i}.txt', 'w') as f:
        f.write('\n'.join(pd.unique(batch)))
ai_patents.to_pickle('../.data/derwent/ccmt_patents.pkl')

ccmt_patents
```
